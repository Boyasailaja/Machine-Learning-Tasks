## Electrical Car price Prediction 

```python
# Step 1: Import required lib 
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
# Step 2: Load the datasets 
df = pd.read_csv("C:\\Users\\yuvra\\Downloads\\ElectricCarData_Clean.csv")
```

```python
# Step 3: Data Explorations 

df.head()
```

```python
df.info()
```

```python
# check the missing value 
df.isnull().sum()
```

```python
# drop columns 
df.drop('Brand',axis= True ,inplace = True)
```

```python
df.drop('Model',axis= True ,inplace = True)
```

```python
df['FastCharge_KmH']=pd.to_numeric(df['FastCharge_KmH'],errors= 'coerce')
```

```python
df['RapidCharge'] = df['RapidCharge'].map({'Yes':1,'No':0})
```

```python
df['PowerTrain'] = df['PowerTrain'].map({'AWD':1,'FWD':2,'RWD':3})
```

```python
df.drop("PlugType",axis = True ,inplace = True)
```

```python
df.drop('BodyStyle',axis  = True , inplace = True)
```

```python
df.drop('Segment',axis = True , inplace = True )
```

```python
df.dtypes
```

```python
# Step 4: Divide datasets into x and y
x = df.drop('PriceEuro',axis = True)
y = df['PriceEuro']
```

```python
# Step 5: spliting model into train and test
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2, random_state = 52)
```

```python
# Step 6: Train the model 
from sklearn.tree import DecisionTreeRegressor

# create the instance of model
DTR = DecisionTreeRegressor()

# train the model
model = DTR.fit(x_train,y_train)

# model prediction
y_pred = model.predict(x_test)
```

```python
# Step 7: model metrics
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error

# check the MSE
MSE = mean_squared_error(y_test,y_pred) 

# check the r2 score
r2_score = r2_score(y_test,y_pred)

# check the MAE
MAE = mean_absolute_error(y_test,y_pred)

print("Mean Squared Error: ",MSE)
print("Mean Absolute Error: ",MAE)
print("R2-Score: ",r2_score)
```
